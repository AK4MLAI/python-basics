```python
# High order function
x = lambda x:x%2==0
y = lambda x:x%2!=0
z = lambda x:x%3==0
```

```python
def return_sum(func, L):
    result = 0
    for i in L:
        if func(i):
            result = result + i
    return result
```

```python
L = [11, 14, 21, 23, 56, 78, 45, 29, 28]
```

```python
print('Sum of even numbers from the give list', return_sum(x, L))
```

```python
print('Sum of odd numbers from the give list', return_sum(y, L))
```

```python
print('Sum of 3 divisable numbers from the give list', return_sum(z, L))
```

# Dictionary comprehension
- create a new dictionary in a concise and readable way.

```python
family = [{"name": "Ambarish", "role": "Engineer"},
    {"name": "Rajni", "role": "Teacher"},
    {"name": "Aditi", "role": "Student"},
    {"name": "Aayushi", "role": "Student"}]
```

```python
mapped = map(lambda p: (p["name"], p["role"]), family)
```

```python
for key,value in mapped:
    print(key, value)
```

```python
# Because mapped is a map iterator, and it was probably already consumed earlier.
DC = {key: value for key, value in mapped if len(key) > 1}
print(DC)
```

```python
mapped = map(lambda p: (p["name"], p["role"]), family)
```

```python
DC = {key: value for key, value in mapped if len(key) > 5}
print(DC)
```

# Class

```python
class Emp:

    # Static/Class Variable
    company_name = "OpenText"

    def __init__(self, emp_id, name, role):
        # Instance variables
        self.emp_id = emp_id
        self.name = name
        self.role = role

    def display(self):
        print(f"ID      : {self.emp_id}")
        print(f"Name    : {self.name}")
        print(f"Role    : {self.role}")
        print(f"Company Name    : {Emp.company_name}")

    # Magic method -> object representation
    def __str__(self):
        return f"Emp({self.emp_id}, {self.name}, {self.role})"

    # Magic method -> object comparison
    def __eq__(self, other):
        return self.emp_id == other.emp_id

    # Magic method -> length using len()
    def __len__(self):
        return len(self.name)

        
# Create Object
e1 = Emp(101, "Ambarish", "Architect")
e2 = Emp(102, "Rajni", "Teacher")

# Access Attributes
print(e1.name)
print(e1.role)

# Call Methods
e1.display()

# __str__
print(e1)

# __eq__
print(e1 == e2)

# __len__
print(len(e1))
```

# Multiple Inheritence
- Python supports mutiple inheritence

```python
class Phone:
    def __init__(self, price, brand, camera):
        print('Inside phone constructor')
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print('Buying a phone')

class Product:
    def buy(self):
        print('Buying a product')

# Whatever is written first, that is taking precedence in the method call, since Phone is first, then buy of Phone will be called
# This is called Method Resolution Order
class SmartPhone(Phone, Product):
    pass

s = SmartPhone(20000, 'Apple', 12)

s.buy()
```

```python
# Python does not support method overloading but with default parameter, it can be achieved
# If you write same method name with different paramter, it will be compiled but at runtime, it will crash asking to provide all the parameters
class Geometry:
    def area(self, a, b=0):
        if(b == 0):
            print('Circle', 3.14*a*a)
        else:
            print('Rectangle', a*b)

obj = Geometry()
obj.area(2)
obj.area(2,3)
```
