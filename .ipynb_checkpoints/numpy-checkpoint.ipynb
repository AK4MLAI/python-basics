# Important NumPy Concepts (Theory)
#### NumPy is the foundation of:
- Data Science
- Machine Learning
- Deep Learning
- Scientific Computing

## Why NumPy Is Important
Compared to normal Python lists, NumPy arrays are:
| Feature                 | Python List | NumPy Array |
| ----------------------- | ----------- | ----------- |
| Speed                   | slower      | very fast   |
| Memory                  | high        | optimized   |
| Vector operations       | limited     | powerful    |
| Multidimensional arrays | difficult   | easy        |


## Core Important Concepts
| Concept                | Importance               |
| ---------------------- | ------------------------ |
| ndarray                | core data structure      |
| Vectorization          | fast computation         |
| Broadcasting           | automatic shape handling |
| Indexing/Slicing       | data extraction          |
| Reshaping              | matrix manipulation      |
| Statistical operations | ML/Data analysis         |
| Linear algebra         | AI/ML foundation         |


## Complete NumPy Example

```python
# ==========================================
# NUMPY COMPLETE EXAMPLE
# Covers almost all important concepts
# ==========================================

import numpy as np


# ==========================================
# 1. CREATE NUMPY ARRAY
# ==========================================

# 1D array
arr1 = np.array([1, 2, 3, 4, 5])

print("1D Array:")
print(arr1)


# 2D array
arr2 = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("\n2D Array:")
print(arr2)


# ==========================================
# 2. ARRAY ATTRIBUTES
# ==========================================

print("\nShape:", arr2.shape)       # rows, columns
print("Dimension:", arr2.ndim)     # number of dimensions
print("Data Type:", arr2.dtype)    # data type
print("Size:", arr2.size)          # total elements


# ==========================================
# 3. SPECIAL ARRAYS
# ==========================================

# Array of zeros
print("\nZeros Array:")
print(np.zeros((2, 3)))

# Array of ones
print("\nOnes Array:")
print(np.ones((2, 2)))

# Identity matrix
print("\nIdentity Matrix:")
print(np.eye(3))

# Range array
print("\nArange:")
print(np.arange(1, 10, 2))

# Evenly spaced values
print("\nLinspace:")
print(np.linspace(0, 1, 5))


# ==========================================
# 4. INDEXING AND SLICING
# ==========================================

print("\nFirst Element:", arr1[0])

print("Last Element:", arr1[-1])

print("Slice:", arr1[1:4])

print("2D Access:", arr2[1][2])


# ==========================================
# 5. ARRAY OPERATIONS
# ==========================================

a = np.array([1, 2, 3])

b = np.array([4, 5, 6])

print("\nAddition:")
print(a + b)

print("Subtraction:")
print(a - b)

print("Multiplication:")
print(a * b)

print("Division:")
print(a / b)


# ==========================================
# 6. VECTORIZATION
# No loops required
# ==========================================

print("\nVectorized Operation:")
print(a * 10)


# ==========================================
# 7. BROADCASTING
# NumPy automatically adjusts shapes
# ==========================================

matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("\nBroadcasting:")
print(matrix + 10)


# ==========================================
# 8. RESHAPING
# ==========================================

arr = np.arange(1, 13)

reshaped = arr.reshape(3, 4)

print("\nReshaped Array:")
print(reshaped)


# ==========================================
# 9. FLATTEN ARRAY
# ==========================================

print("\nFlatten:")
print(reshaped.flatten())


# ==========================================
# 10. STATISTICAL OPERATIONS
# ==========================================

nums = np.array([10, 20, 30, 40, 50])

print("\nSum:", np.sum(nums))

print("Mean:", np.mean(nums))

print("Max:", np.max(nums))

print("Min:", np.min(nums))

print("Standard Deviation:", np.std(nums))


# ==========================================
# 11. BOOLEAN FILTERING
# ==========================================

print("\nFilter Greater Than 25:")
print(nums[nums > 25])


# ==========================================
# 12. ITERATING ARRAY
# ==========================================

print("\nIterating:")

for value in nums:
    print(value)


# ==========================================
# 13. COPY VS VIEW
# ==========================================

original = np.array([1, 2, 3])

view = original.view()

copy = original.copy()

original[0] = 100

print("\nOriginal:", original)

# View changes with original
print("View:", view)

# Copy remains independent
print("Copy:", copy)


# ==========================================
# 14. RANDOM NUMBERS
# ==========================================

print("\nRandom Integers:")
print(np.random.randint(1, 10, 5))

print("\nRandom Matrix:")
print(np.random.rand(2, 3))


# ==========================================
# 15. MATRIX OPERATIONS
# ==========================================

m1 = np.array([
    [1, 2],
    [3, 4]
])

m2 = np.array([
    [5, 6],
    [7, 8]
])

print("\nMatrix Multiplication:")
print(np.matmul(m1, m2))


# ==========================================
# 16. TRANSPOSE MATRIX
# ==========================================

print("\nTranspose:")
print(m1.T)


# ==========================================
# 17. CONDITIONAL OPERATIONS
# ==========================================

print("\nWhere Condition:")
print(np.where(nums > 25, "High", "Low"))


# ==========================================
# 18. SAVE AND LOAD ARRAY
# ==========================================

np.save("data.npy", nums)

loaded = np.load("data.npy")

print("\nLoaded Array:")
print(loaded)


# ==========================================
# IMPORTANT NOTES
# ==========================================

"""
NumPy arrays are:
- faster than Python lists
- memory optimized
- support vectorized operations

Broadcasting:
Allows operations between arrays of different shapes.

Vectorization:
Perform operations without loops.

ndarray:
Core NumPy data structure.
"""
```
