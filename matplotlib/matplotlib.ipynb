- scientific plotting

# Complete Matplotlib + Pandas Example Using Titanic Dataset

## Important Columns
| Column     | Meaning         |
| ---------- | --------------- |
| `Survived` | 0 = No, 1 = Yes |
| `Pclass`   | passenger class |
| `Sex`      | male/female     |
| `Age`      | age             |
| `Fare`     | ticket fare     |
| `Embarked` | boarding port   |


## Complete Example

```python
# ==========================================
# COMPLETE MATPLOTLIB + PANDAS EXAMPLE
# USING TITANIC DATASET
# ==========================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


# ==========================================
# 1. LOAD CSV FILE
# ==========================================

df = pd.read_csv("titanic.csv")

print(df.head())


# ==========================================
# 2. BASIC DATA ANALYSIS
# ==========================================

print("\nShape:")
print(df.shape)

print("\nColumns:")
print(df.columns)

print("\nMissing Values:")
print(df.isnull().sum())


# ==========================================
# 3. HANDLE MISSING VALUES
# ==========================================

# Fill missing Age with mean age
df["Age"] = df["Age"].fillna(df["Age"].mean())


# ==========================================
# 4. SIMPLE LINE PLOT
# ==========================================

# Sort by PassengerId
df_sorted = df.sort_values(by="PassengerId")

plt.figure(figsize=(8,5))

plt.plot(
    df_sorted["PassengerId"].head(50),
    df_sorted["Fare"].head(50)
)

plt.title("Passenger Fare Trend")

plt.xlabel("Passenger ID")

plt.ylabel("Fare")

plt.grid(True)

plt.show()


# ==========================================
# 5. BAR CHART
# Survival Count
# ==========================================

survival_counts = df["Survived"].value_counts()

plt.figure(figsize=(6,4))

plt.bar(
    ["Not Survived", "Survived"],
    survival_counts
)

plt.title("Survival Count")

plt.ylabel("Passengers")

plt.show()


# ==========================================
# 6. HISTOGRAM
# Age Distribution
# ==========================================

plt.figure(figsize=(8,5))

plt.hist(
    df["Age"],
    bins=20
)

plt.title("Age Distribution")

plt.xlabel("Age")

plt.ylabel("Frequency")

plt.show()


# ==========================================
# 7. SCATTER PLOT
# Fare vs Age
# ==========================================

plt.figure(figsize=(8,5))

plt.scatter(
    df["Age"],
    df["Fare"]
)

plt.title("Age vs Fare")

plt.xlabel("Age")

plt.ylabel("Fare")

plt.show()


# ==========================================
# 8. PIE CHART
# Gender Distribution
# ==========================================

gender_counts = df["Sex"].value_counts()

plt.figure(figsize=(6,6))

plt.pie(
    gender_counts,
    labels=gender_counts.index,
    autopct="%1.1f%%"
)

plt.title("Gender Distribution")

plt.show()


# ==========================================
# 9. MULTIPLE LINE PLOTS
# ==========================================

male_survival = df[df["Sex"] == "male"]["Survived"].value_counts()

female_survival = df[df["Sex"] == "female"]["Survived"].value_counts()

plt.figure(figsize=(8,5))

plt.plot(
    male_survival.index,
    male_survival.values,
    marker="o",
    label="Male"
)

plt.plot(
    female_survival.index,
    female_survival.values,
    marker="o",
    label="Female"
)

plt.title("Survival Comparison")

plt.xlabel("Survival")

plt.ylabel("Count")

plt.legend()

plt.grid(True)

plt.show()


# ==========================================
# 10. SUBPLOTS
# ==========================================

plt.figure(figsize=(12,5))


# First subplot
plt.subplot(1,2,1)

plt.hist(df["Age"], bins=20)

plt.title("Age Distribution")


# Second subplot
plt.subplot(1,2,2)

plt.hist(df["Fare"], bins=20)

plt.title("Fare Distribution")

plt.show()


# ==========================================
# 11. BOXPLOT
# Detect outliers
# ==========================================

plt.figure(figsize=(6,4))

plt.boxplot(df["Fare"])

plt.title("Fare Boxplot")

plt.show()


# ==========================================
# 12. PANDAS BUILT-IN PLOTTING
# ==========================================

# Pandas internally uses matplotlib

df["Pclass"].value_counts().plot(
    kind="bar"
)

plt.title("Passenger Class Count")

plt.show()


# ==========================================
# 13. GROUP BY + PLOT
# ==========================================

avg_fare = df.groupby("Pclass")["Fare"].mean()

avg_fare.plot(
    kind="bar"
)

plt.title("Average Fare by Passenger Class")

plt.xlabel("Passenger Class")

plt.ylabel("Average Fare")

plt.show()


# ==========================================
# 14. NUMPY + MATPLOTLIB
# Mathematical Plot
# ==========================================

x = np.linspace(0, 10, 100)

y = np.sin(x)

plt.figure(figsize=(8,5))

plt.plot(x, y)

plt.title("Sine Wave")

plt.xlabel("X")

plt.ylabel("sin(X)")

plt.grid(True)

plt.show()


# ==========================================
# 15. CUSTOMIZE STYLE
# ==========================================

plt.figure(figsize=(8,5))

plt.plot(
    x,
    y,
    color="red",
    linestyle="--",
    linewidth=2,
    marker="o",
    markersize=4
)

plt.title("Styled Sine Wave")

plt.grid(True)

plt.show()


# ==========================================
# 16. SAVE GRAPH
# ==========================================

plt.figure(figsize=(6,4))

plt.hist(df["Age"])

plt.title("Age Histogram")

plt.savefig("age_histogram.png")

print("\nGraph Saved")


# ==========================================
# IMPORTANT CONCEPTS
# ==========================================

"""
Matplotlib:
Core plotting library

Pandas:
Uses matplotlib internally

NumPy:
Used for numerical computations

Most Important Plot Types:
- line plot
- bar chart
- scatter plot
- histogram
- pie chart
- boxplot

Subplots:
Multiple graphs in one figure

figure():
Creates plotting canvas

grid():
Adds grid lines

legend():
Shows labels
"""
```
