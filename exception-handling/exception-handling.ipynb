```python
# ==========================================
# COMPILE-TIME / SYNTAX ERRORS IN PYTHON
# ==========================================

# NOTE:
# Python mainly has:
# 1. Syntax Errors (during parsing/compilation)
# 2. Runtime Exceptions

# Python does NOT have compile-time checking
# like Java/C++.

# These examples cause SyntaxError before execution.


# ------------------------------------------
# 1. MISSING COLON
# ------------------------------------------

# if True
#     print("Hello")

# Error:
# SyntaxError: expected ':'


# ------------------------------------------
# 2. INVALID INDENTATION
# ------------------------------------------

# def test():
# print("Hello")

# Error:
# IndentationError


# ------------------------------------------
# 3. UNMATCHED PARENTHESES
# ------------------------------------------

# print("Hello"

# Error:
# SyntaxError: '(' was never closed


# ------------------------------------------
# 4. INVALID VARIABLE NAME
# ------------------------------------------

# 1name = "Ambarish"

# Error:
# SyntaxError: invalid decimal literal


# ------------------------------------------
# 5. USING RESERVED KEYWORD
# ------------------------------------------

# class = "Python"

# Error:
# SyntaxError: invalid syntax


# ------------------------------------------
# 6. INVALID ASSIGNMENT
# ------------------------------------------

# 10 = x

# Error:
# SyntaxError: cannot assign to literal


# ------------------------------------------
# 7. MISSING QUOTES
# ------------------------------------------

# name = "Ambarish

# Error:
# SyntaxError: unterminated string literal


# ------------------------------------------
# 8. INVALID FUNCTION DEFINITION
# ------------------------------------------

# def():
#     pass

# Error:
# SyntaxError: invalid syntax


# ------------------------------------------
# 9. DUPLICATE ARGUMENTS
# ------------------------------------------

# def add(a, a):
#     return a

# Error:
# SyntaxError: duplicate argument


# ------------------------------------------
# 10. RETURN OUTSIDE FUNCTION
# ------------------------------------------

# return 10

# Error:
# SyntaxError: 'return' outside function


# ------------------------------------------
# 11. BREAK OUTSIDE LOOP
# ------------------------------------------

# break

# Error:
# SyntaxError: 'break' outside loop


# ------------------------------------------
# 12. CONTINUE OUTSIDE LOOP
# ------------------------------------------

# continue

# Error:
# SyntaxError: 'continue' not properly in loop


# ------------------------------------------
# 13. INVALID IMPORT
# ------------------------------------------

# import 123module

# Error:
# SyntaxError


# ------------------------------------------
# 14. INVALID FOR LOOP
# ------------------------------------------

# for i range(5):
#     print(i)

# Error:
# SyntaxError


# ------------------------------------------
# 15. INVALID IF STATEMENT
# ------------------------------------------

# if x == 10
#     print(x)

# Error:
# SyntaxError: expected ':'


# ------------------------------------------
# 16. WRONG INDENTATION LEVEL
# ------------------------------------------

# if True:
#     print("Hello")
#       print("World")

# Error:
# IndentationError


# ------------------------------------------
# 17. INVALID LIST SYNTAX
# ------------------------------------------

# nums = [1, 2, 3

# Error:
# SyntaxError


# ------------------------------------------
# 18. INVALID DICTIONARY SYNTAX
# ------------------------------------------

# d = {"name" "Ambarish"}

# Error:
# SyntaxError


# ------------------------------------------
# 19. INVALID TRY-EXCEPT
# ------------------------------------------

# try:
#     print("Hello")

# Error:
# SyntaxError
# because except/finally missing


# ------------------------------------------
# 20. INVALID CLASS DEFINITION
# ------------------------------------------

# class Emp
#     pass

# Error:
# SyntaxError: expected ':'


# ==========================================
# IMPORTANT CONCEPT
# ==========================================

"""
Compile-time errors in Python are mostly:

1. SyntaxError
2. IndentationError
3. TabError

These errors occur BEFORE program execution.
Python parser/interpreter detects them first.
"""


# ==========================================
# DIFFERENCE
# ==========================================

"""
Syntax Error:
    Detected before execution

Runtime Error:
    Occurs during execution

Example Runtime Errors:
    ZeroDivisionError
    FileNotFoundError
    ValueError
"""
```

```python
# ==========================================
# EXCEPTION HANDLING IN PYTHON
# Covers almost all important concepts
# ==========================================


# ------------------------------------------
# 1. BASIC TRY-EXCEPT
# ------------------------------------------

try:

    num = 10 / 0

except ZeroDivisionError:

    print("Cannot divide by zero")


# ------------------------------------------
# 2. MULTIPLE EXCEPT BLOCKS
# ------------------------------------------

try:

    number = int("ABC")

except ValueError:

    print("Invalid number format")

except ZeroDivisionError:

    print("Division by zero occurred")


# ------------------------------------------
# 3. GENERIC EXCEPTION
# Handles any exception
# ------------------------------------------

try:

    data = [1, 2, 3]

    print(data[10])

except Exception as e:

    print("Exception occurred:", e)


# ------------------------------------------
# 4. ELSE BLOCK
# Executes only if NO exception occurs
# ------------------------------------------

try:

    result = 10 / 2

except ZeroDivisionError:

    print("Error")

else:

    print("Division successful:", result)


# ------------------------------------------
# 5. FINALLY BLOCK
# Always executes
# Used for cleanup
# ------------------------------------------

try:

    file = open("demo.txt", "w")

    file.write("Hello")

except Exception as e:

    print(e)

finally:

    print("Finally block executed")

    file.close()


# ------------------------------------------
# 6. USER-DEFINED EXCEPTION
# ------------------------------------------

class InvalidAgeException(Exception):

    pass


try:

    age = -5

    if age < 0:
        raise InvalidAgeException("Age cannot be negative")

except InvalidAgeException as e:

    print("Custom Exception:", e)


# ------------------------------------------
# 7. RAISE EXCEPTION MANUALLY
# ------------------------------------------

try:

    amount = 500

    if amount < 1000:

        raise ValueError("Minimum amount should be 1000")

except ValueError as e:

    print(e)


# ------------------------------------------
# 8. MULTIPLE EXCEPTIONS IN SINGLE BLOCK
# ------------------------------------------

try:

    value = int("ABC")

except (ValueError, TypeError) as e:

    print("Handled Multiple Exceptions:", e)


# ------------------------------------------
# 9. NESTED TRY-EXCEPT
# ------------------------------------------

try:

    try:

        num = 10 / 0

    except ZeroDivisionError:

        print("Inner try handled division error")

except Exception:

    print("Outer try")


# ------------------------------------------
# 10. FILE HANDLING EXCEPTION
# ------------------------------------------

try:

    with open("missing.txt", "r") as file:

        print(file.read())

except FileNotFoundError:

    print("File does not exist")


# ------------------------------------------
# 11. ASSERTION EXCEPTION
# ------------------------------------------

try:

    age = -1

    assert age >= 0, "Age must be positive"

except AssertionError as e:

    print(e)


# ------------------------------------------
# 12. USING finally FOR RESOURCE CLEANUP
# ------------------------------------------

try:

    print("Database connection opened")

    x = 10 / 2

except Exception as e:

    print(e)

finally:

    print("Database connection closed")


# ------------------------------------------
# 13. COMPLETE FLOW DEMO
# try -> except -> else -> finally
# ------------------------------------------

try:

    num = int(input("Enter number: "))

    result = 100 / num

except ValueError:

    print("Please enter valid integer")

except ZeroDivisionError:

    print("Cannot divide by zero")

else:

    print("Result:", result)

finally:

    print("Execution completed")


# ==========================================
# IMPORTANT EXCEPTION HIERARCHY
# ==========================================

"""
BaseException
    Exception
        ArithmeticError
            ZeroDivisionError

        ValueError
        TypeError
        FileNotFoundError
        AssertionError
"""


# ==========================================
# IMPORTANT KEYWORDS
# ==========================================

"""
try      -> risky code
except   -> handle exception
else     -> executes if no exception
finally  -> always executes
raise    -> manually throw exception
assert   -> validate condition
"""


# ==========================================
# BEST PRACTICES
# ==========================================

"""
1. Catch specific exceptions
2. Avoid bare except:
3. Use finally for cleanup
4. Prefer with for file handling
5. Create custom exceptions when needed
"""
```
