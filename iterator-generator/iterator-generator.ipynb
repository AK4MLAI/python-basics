# Simple Iterator Example

## An iterator is an object that:
- remembers current position
- returns one item at a time

```python
# List is iterable
numbers = [10, 20, 30]

# Convert iterable into iterator
itr = iter(numbers)

# Fetch items one by one
print(next(itr))   # 10
print(next(itr))   # 20
print(next(itr))   # 30

# Next call raises StopIteration
# print(next(itr))
```

## Important Functions
| Function | Purpose         |
| -------- | --------------- |
| `iter()` | create iterator |
| `next()` | get next value  |


# Custom Iterator Example
## A custom iterator class must implement:
### `__iter__()`
### `__next__()`

```python
class Counter:

    def __init__(self, start, end):

        self.current = start
        self.end = end

    # Returns iterator object itself
    def __iter__(self):
        return self

    # Returns next value
    def __next__(self):

        if self.current > self.end:
            raise StopIteration

        value = self.current

        self.current += 1

        return value


# Create iterator object
c = Counter(1, 5)

# Iterate using for loop
for num in c:
    print(num)
```

## How It Works Internally
### for loop internally does:
#### 
itr = iter(c)

while True:
    value = next(itr)

until:

StopIteration    

# Simple Generator Example
## Generator is an easier way to create iterators.
### Uses: yield
#### instead of:
- `__iter__`
- `__next__`


```python
def simple_generator():

    yield 1
    yield 2
    yield 3


g = simple_generator()

print(next(g))
print(next(g))
print(next(g))
```

## Important Concept

### yield pauses function and remembers state.

### Generator Using Loop

```python
def number_generator():

    for i in range(1, 6):
        yield i


g = number_generator()

for num in g:
    print(num)
```

# Custom Generator Example

```python
# Generate square numbers lazily.
def square_generator(limit):

    for i in range(1, limit + 1):

        # Yield square value
        yield i * i


g = square_generator(5)

for value in g:
    print(value)
```

### Why Generators Are Powerful

#### Generators:
- produce values one at a time
- memory efficient
- lazy evaluation

## Normal List vs Generator

```python
# List
nums = [x*x for x in range(1000000)]
# Creates ALL values in memory.
```

```python
# Generator
nums = (x*x for x in range(1000000))
# Creates values only when needed.
```

### Generator Expression

```python
g = (x * x for x in range(5))

print(next(g))
print(next(g))
```

# Iterator vs Generator
| Iterator                         | Generator       |
| -------------------------------- | --------------- |
| Need class                       | Simple function |
| Implement `__iter__`, `__next__` | Use `yield`     |
| More code                        | Less code       |
| Manual state management          | Automatic state |
