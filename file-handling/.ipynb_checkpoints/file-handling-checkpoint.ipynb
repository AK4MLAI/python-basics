# Example WITHOUT with

```python
# Open file manually
file = open("demo.txt", "w")

# Write content
file.write("Hello Ambarish")

# Must close manually
file.close()

print("File closed manually")
```

## Problem Without with
### Suppose exception occurs before:
__file.close()__
### Then file may remain open.
- file not closed properly
- memory/resource leak possible

# Example WITH with

```python
with open("demo.txt", "w") as file:

    file.write("Hello Ambarish")

print("File closed automatically")
```

## Main Benefit of with
### Even if exception occurs:
### Python automatically closes file safely.

#### Internally
with uses something called:
- __Context Manager__

It automatically handles:
- setup
- cleanup

## Text File Oeration

```python
# ==========================================
# FILE HANDLING EXAMPLE IN PYTHON
# ==========================================

import os


# ------------------------------------------
# 1. WRITE TO FILE (w)
# Creates file if not exists
# Overwrites existing content
# ------------------------------------------

with open("sample.txt", "w") as file:

    file.write("Hello Ambarish\n")
    file.write("Welcome to Python File Handling\n")

print("File written successfully")


# ------------------------------------------
# 2. APPEND TO FILE (a)
# Adds content at end of file
# ------------------------------------------

with open("sample.txt", "a") as file:

    file.write("Appending new line\n")

print("Content appended")


# ------------------------------------------
# 3. READ ENTIRE FILE (r)
# ------------------------------------------

with open("sample.txt", "r") as file:

    content = file.read()

    print("\nFull File Content:")
    print(content)


# ------------------------------------------
# 4. READ LINE BY LINE
# ------------------------------------------

with open("sample.txt", "r") as file:

    print("\nReading line by line:")

    for line in file:
        print(line.strip())


# ------------------------------------------
# 5. READ SINGLE LINE
# ------------------------------------------

with open("sample.txt", "r") as file:

    first_line = file.readline()

    print("\nFirst Line:")
    print(first_line)


# ------------------------------------------
# 6. READ ALL LINES INTO LIST
# ------------------------------------------

with open("sample.txt", "r") as file:

    lines = file.readlines()

    print("\nLines List:")
    print(lines)


# ------------------------------------------
# 7. CHECK FILE EXISTS
# ------------------------------------------

if os.path.exists("sample.txt"):

    print("\nFile exists")


# ------------------------------------------
# 8. GET FILE DETAILS
# ------------------------------------------

size = os.path.getsize("sample.txt")

print("\nFile Size:", size, "bytes")


# ------------------------------------------
# 9. RENAME FILE
# ------------------------------------------

os.rename("sample.txt", "new_sample.txt")

print("\nFile renamed")


# ------------------------------------------
# 10. DELETE FILE
# ------------------------------------------

os.remove("new_sample.txt")

print("File deleted")


# ==========================================
# FILE MODES SUMMARY
# ==========================================

"""
r  -> Read
w  -> Write (overwrite)
a  -> Append
x  -> Create new file
rb -> Read binary
wb -> Write binary
r+ -> Read + Write
"""


# ==========================================
# EXCEPTION HANDLING
# ==========================================

try:

    with open("missing.txt", "r") as file:

        print(file.read())

except FileNotFoundError:

    print("\nFile not found error handled")
```

## Binary File Operation

```python
# ==========================================
# BINARY FILE HANDLING EXAMPLE
# ==========================================

# Write binary data to file
with open("sample.bin", "wb") as file:

    # Binary data
    data = b"Hello Ambarish"

    file.write(data)

print("Binary file written successfully")


# Read binary data from file
with open("sample.bin", "rb") as file:

    content = file.read()

    print("Binary Content:")
    print(content)
```

## Important Modes
| Mode | Meaning      |
| ---- | ------------ |
| `wb` | write binary |
| `rb` | read binary  |


## Difference Between Text and Binary
| Text Mode          | Binary Mode      |
| ------------------ | ---------------- |
| `"w"`              | `"wb"`           |
| `"r"`              | `"rb"`           |
| works with strings | works with bytes |


### Simple Image Copy Example

```python
# Copy image using binary mode

with open("input.jpg", "rb") as source:

    data = source.read()

with open("copy.jpg", "wb") as target:

    target.write(data)

print("Image copied")
```
