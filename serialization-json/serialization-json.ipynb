# 1. Serialization and Deserialization Using Pickle

### In Python:

- __Serialization = convert object → byte stream/file__
- __Deserialization = convert file/bytes → object__

Using:
__pickle__


## Pickle Example

```python
import pickle


# Python object
employee = {
    "id": 101,
    "name": "Ambarish",
    "role": "Architect"
}


# ------------------------------------------
# SERIALIZATION
# Object -> Binary File
# ------------------------------------------

with open("employee.pkl", "wb") as file:

    pickle.dump(employee, file)

print("Object serialized")


# ------------------------------------------
# DESERIALIZATION
# Binary File -> Object
# ------------------------------------------

with open("employee.pkl", "rb") as file:

    data = pickle.load(file)

print("Deserialized Object:")
print(data)
```

### Important Methods
| Method          | Purpose     |
| --------------- | ----------- |
| `pickle.dump()` | serialize   |
| `pickle.load()` | deserialize |


#### Important Note

Pickle stores:

- __Python-specific binary format__

Not human readable.

# 2. JSON Example
### JSON is:
- __lightweight__
- __human-readable__
- __language-independent__

Using:
- __json__

## JSON Serialization

```python
import json


employee = {
    "id": 101,
    "name": "Ambarish",
    "role": "Architect"
}


# ------------------------------------------
# PYTHON OBJECT -> JSON FILE
# ------------------------------------------

with open("employee.json", "w") as file:

    json.dump(employee, file, indent=4)

print("JSON written")
```

## JSON Deserialization

```python
import json


with open("employee.json", "r") as file:

    data = json.load(file)

print(data)
```

# 3. CSV Example
CSV = Comma Separated Values

Using:
- __csv__

## Write CSV

```python
import csv


rows = [
    ["ID", "Name", "Role"],
    [101, "Ambarish", "Architect"],
    [102, "Rajni", "Teacher"]
]


with open("employees.csv", "w", newline="") as file:

    writer = csv.writer(file)

    writer.writerows(rows)

print("CSV written")
```

## Read CSV

```python
import csv


with open("employees.csv", "r") as file:

    reader = csv.reader(file)

    for row in reader:
        print(row)
```

# 4. YAML Example
#### YAML is:
- human-friendly
- configuration-focused

Common in:
- Kubernetes
- Docker
- CI/CD
- AI configs

Using:
- __pyyaml__

```python
pip install pyyaml
```

## Write YAML

```python
import yaml


employee = {
    "id": 101,
    "name": "Ambarish",
    "role": "Architect"
}


with open("employee.yaml", "w") as file:

    yaml.dump(employee, file)

print("YAML written")
```

## Read YAML

```python
import yaml


with open("employee.yaml", "r") as file:

    data = yaml.safe_load(file)

print(data)
```

## Comparison
| Format | Human Readable | Language Independent | Binary |
| ------ | -------------- | -------------------- | ------ |
| Pickle | ❌              | ❌                    | ✅      |
| JSON   | ✅              | ✅                    | ❌      |

## When To Use
| Use Case                  | Recommended |
| ------------------------- | ----------- |
| Python object persistence | Pickle      |
| APIs                      | JSON        |
| Excel/tabular data        | CSV         |
| Configurations            | YAML        |


## Important Security Note

Never load untrusted pickle files because:
- pickle can execute arbitrary code

JSON/YAML are safer for external data exchange.
