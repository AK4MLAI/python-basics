# Operators
- Arithmetic Operators
- Comparison Operators
- Logical Operators
- Bitwise Operators
- Assignment Operators
- Identity Operators
- Membership Operators

```python
# Arithmetic Operators
x = 5
y = 2
print(x + y)
print(x/y)
print(x%y)
print(x**y) # power of
print(x//2) # Integer division - convert float to integer
```

```python
# Comparison Operators
print(x >= y)
print(x <= y)
print(x == y)
print(x != y)
```

```python
# Logical Operator
print(x or y)
print(x and y)
print(not x)
```

```python
# Bitwise operators
x = 2
y = 3
print(x & y)
print(x | y)
print(x >> 2)
print(y << 3)
print(~x)
```

```python
# Assignment Operators
a = 3
print(a)
a += 3
print(a)
a *= 3
print(a)
# ++aa and a++ is not supported in python
```

```python
# Identity Operators
a = 3
b = 3
print(a is b) # Same memory location
a = 'Hello'
b = 'Hello'
print(a is b)
a = [1,2,3]
b = [1,2,3]
print(a is b)
```

```python
# Membership Operators
x = 'Deoghar'
print('D' in x)
print('D' not in x)
x = [1,2,3]
print(3 in x)
print(5 in x)
```

| Method          | Purpose         |
| --------------- | --------------- |
| `dict.keys()`   | keys            |
| `dict.values()` | values          |
| `dict.items()`  | key-value pairs |


```python
person = {'Name': 'Ambarish', 'Age': 40}
for i in person:
    print(i)
```

```python
for key in person.keys():
    print(key)
```

```python
for value in person.values():
    print(value)
```

```python
for key, value in person.items():
    print(key, ':', value)
```

```python
# Formatting
for key, value in person.items():
    print(f"{key} : {value}")
```

```python
employees = {
    101: {"name": "John", "role": "Dev"},
    102: {"name": "Alice", "role": "QA"}
}

for emp_id, details in employees.items():
    print(emp_id, details["name"])
```

# Set in Python
- Set does not allow duplicates
- Set does not support indexing or slicing
- Set does not store value in order as given because it creates hashing and it orders the values based on hashing
- Set does not allow to hold mutable data types
- Set itself is mutable data type, so it set inside set is not possible due to rule 3 and therefore it cannot have 2D sets

```python
s1 = {}
print(s1)
print(type(s1)) # By default, {} is treated as dictionary
s2 = set()
print(s2)
print(type(s2))
s3 = {2,2.2,'Hello',2+4j,True}
print(s3)
s4 = {1,2.2,'Hello',2+4j,True} # True is not print because True is nothing but 1 and 1 is already there in set and set does not allow duplicates
print(s4)
s5 = {1,1,2,2,3,4,5,4}
print(s5)
s6 = {(1,2,3),'Hello'} #Stoes immutable tuples and string
print(s6)
#print(s6[0]) #THis will not work because it does store value based on index, it stores the value based on hashing
s7 = {[1,2,3],'Hello'} #This will give error because list is mutable
print(s7)
```

```python
#Edit is not possible
s8 = {1,2,3,4,5}
s8[0]=100
print(s8)
```

```python
# We create set and then convert into the list, so that we can edit that list and then convert back that list into set
# But basically we are not editing the same set, It creates new set, hence set is not editable
s9 = {1,2,3,4,5}
print(s9)
print(id(s9)) # different memory location
l9 = list(s9)
print(l9)
l9[0]=100
print(l9)
s9 = set(l9)
print(s9)
print(id(s9)) # different memory location
```

```python
s10 = {1, 2, 3, 4, 5}
print(s10)
s10.add(6)
print(s10)
#del s1
s10.remove(3) # Remove item from the set
print(s10)
s10.pop()
print(s10)
print(2 in s10)
print(len(s10))
```

```python
s11 = {1,2,3,4,5}
s12 = {4,5,6,7,8}
#s11 + s12 #concatenation is not allowed in set
print(s11.union(s12))
print(s11.intersection(s12))
print(s11.difference(s12))
print(s11.symmetric_difference(s12))
print(s11.isdisjoint(s2))
print(s11.issubset(s12))
print(s11.issuperset(s12))
```

# Dictionary in Python
- Dictionary is like a map
- no indexing supported
- key must be unique and immutable
- value can be mutable
- Dictionary is mutable types
- key is based on hashing

```python
d1 = {}
print(d1)
d2 = {'Name': 'Ambarish', 'Gender': 'Male'}
print(d2)
d3 = {'Name':'Ambarish', 'Marks':{'Math':97, 'Science':94}}
print(d3)
#d2[0] # does not support indexing
print(d2['Name'])
print(d3['Marks']['Math'])
d2['Name']='Rajni'
d2['Gender']='Female'
print(d2)
print(d2.get('Name'))
d2['Age']=40
print(d2)
del d2['Age']
print(d2)
```

```python
for i in d3:
    print(i,d3[i])
```
