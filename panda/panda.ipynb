# Important Pandas Concepts (Theory)
pandas is the most important library for:
- Data Analysis
- CSV Processing
- Data Cleaning
- Machine Learning preprocessing
- Reporting

Built on top of:
- NumPy

# Core Data Structures
| Structure   | Description               |
| ----------- | ------------------------- |
| `Series`    | 1D labeled array          |
| `DataFrame` | 2D table (rows + columns) |


## Why Pandas Is Important
Without pandas:
- CSV handling is difficult
- filtering is manual
- grouping requires loops

Pandas provides:
- SQL-like operations
- Excel-like operations
- powerful data manipulation

## Complete Pandas Example

```python
# ==========================================
# PANDAS COMPLETE EXAMPLE
# Covers almost all important concepts
# ==========================================

import pandas as pd


# ==========================================
# 1. READ CSV FILE
# ==========================================

# Load CSV into DataFrame
df = pd.read_csv("employees.csv")

print("DataFrame:")
print(df)


# ==========================================
# 2. BASIC ATTRIBUTES
# ==========================================

print("\nRows and Columns:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nTotal Elements:")
print(df.size)

print("\nDimension:")
print(df.ndim)

print("\nIndex:")
print(df.index)


# ==========================================
# 3. BASIC FUNCTIONS
# ==========================================

# First 5 rows
print("\nHead:")
print(df.head())

# Last 5 rows
print("\nTail:")
print(df.tail())

# Statistical summary
print("\nDescribe:")
print(df.describe())

# Information about dataframe
print("\nInfo:")
print(df.info())


# ==========================================
# 4. SELECT COLUMNS
# ==========================================

print("\nSingle Column:")
print(df["Name"])

print("\nMultiple Columns:")
print(df[["Name", "Salary"]])


# ==========================================
# 5. ACCESS ROWS
# ==========================================

# Using index location
print("\nFirst Row:")
print(df.iloc[0])

# Using label
print("\nRows 1 to 3:")
print(df.loc[1:3])


# ==========================================
# 6. FILTERING DATA
# ==========================================

# Employees with salary > 90000
high_salary = df[df["Salary"] > 90000]

print("\nHigh Salary Employees:")
print(high_salary)


# Multiple conditions
filtered = df[
    (df["Department"] == "AI") &
    (df["Salary"] > 100000)
]

print("\nFiltered Data:")
print(filtered)


# ==========================================
# 7. SORTING
# ==========================================

print("\nSort by Salary:")
print(df.sort_values(by="Salary"))

print("\nSort Descending:")
print(df.sort_values(by="Salary", ascending=False))


# ==========================================
# 8. ADD NEW COLUMN
# ==========================================

# Bonus = 10% of salary
df["Bonus"] = df["Salary"] * 0.10

print("\nAdded Bonus Column:")
print(df)


# ==========================================
# 9. UPDATE VALUES
# ==========================================

df.loc[0, "Salary"] = 160000

print("\nUpdated Salary:")
print(df)


# ==========================================
# 10. DELETE COLUMN
# ==========================================

temp_df = df.drop("Bonus", axis=1)

print("\nAfter Dropping Bonus:")
print(temp_df)


# ==========================================
# 11. HANDLE MISSING VALUES
# ==========================================

# Check missing values
print("\nMissing Values:")
print(df.isnull())

# Count missing values
print("\nMissing Count:")
print(df.isnull().sum())


# ==========================================
# 12. GROUP BY
# Similar to SQL GROUP BY
# ==========================================

grouped = df.groupby("Department")["Salary"].mean()

print("\nAverage Salary by Department:")
print(grouped)


# ==========================================
# 13. UNIQUE VALUES
# ==========================================

print("\nUnique Departments:")
print(df["Department"].unique())


# ==========================================
# 14. VALUE COUNTS
# ==========================================

print("\nDepartment Counts:")
print(df["Department"].value_counts())


# ==========================================
# 15. STRING OPERATIONS
# ==========================================

print("\nUppercase Names:")
print(df["Name"].str.upper())


# ==========================================
# 16. APPLY FUNCTION
# ==========================================

def categorize_salary(salary):

    if salary > 100000:
        return "High"

    return "Medium"


df["Category"] = df["Salary"].apply(categorize_salary)

print("\nSalary Category:")
print(df)


# ==========================================
# 17. ITERATE DATAFRAME
# ==========================================

print("\nIterating Rows:")

for index, row in df.iterrows():

    print(row["Name"], row["Salary"])


# ==========================================
# 18. SAVE TO CSV
# ==========================================

df.to_csv("updated_employees.csv", index=False)

print("\nCSV Saved")


# ==========================================
# 19. CREATE DATAFRAME MANUALLY
# ==========================================

data = {
    "Name": ["A", "B"],
    "Age": [25, 30]
}

df2 = pd.DataFrame(data)

print("\nManual DataFrame:")
print(df2)


# ==========================================
# 20. BOOLEAN CONDITIONS
# ==========================================

print("\nEmployees from India:")
print(df[df["Country"] == "India"])


# ==========================================
# IMPORTANT NOTES
# ==========================================

"""
DataFrame:
2D tabular structure in pandas

Series:
Single column in pandas

iloc:
Integer-location based indexing

loc:
Label-based indexing

groupby:
Aggregation similar to SQL GROUP BY

apply():
Apply custom function to column
"""
```

## Complete Pandas Example Using Titanic Dataset

### Data Dictionary

| Variable | Definition | Key |
|---|---|---|
| survival | Survival | 0 = No, 1 = Yes |
| pclass | Ticket class | 1 = 1st, 2 = 2nd, 3 = 3rd |
| sex | Sex | |
| age | Age in years | |
| sibsp | # of siblings / spouses aboard the Titanic | |
| parch | # of parents / children aboard the Titanic | |
| ticket | Ticket number | |
| fare | Passenger fare | |
| cabin | Cabin number | |
| embarked | Port of Embarkation | C = Cherbourg, Q = Queenstown, S = Southampton |

```python
# ==========================================
# PANDAS USING REAL TITANIC DATASET
# ==========================================

import pandas as pd


# ==========================================
# 1. LOAD CSV FILE
# ==========================================

df = pd.read_csv("titanic.csv")

print(df)


# ==========================================
# 2. BASIC ATTRIBUTES
# ==========================================

print("\nShape:")
print(df.shape)

print("\nColumns:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nInfo:")
print(df.info())


# ==========================================
# 3. FIRST / LAST ROWS
# ==========================================

print("\nHead:")
print(df.head())

print("\nTail:")
print(df.tail())


# ==========================================
# 4. STATISTICAL SUMMARY
# ==========================================

print("\nDescribe:")
print(df.describe())


# ==========================================
# 5. SELECT COLUMNS
# ==========================================

print("\nNames:")
print(df["Name"])

print("\nName and Age:")
print(df[["Name", "Age"]])


# ==========================================
# 6. ACCESS ROWS
# ==========================================

print("\nFirst Row:")
print(df.iloc[0])

print("\nRows 1 to 5:")
print(df.iloc[1:6])


# ==========================================
# 7. FILTERING
# ==========================================

# Female passengers
females = df[df["Sex"] == "female"]

print("\nFemale Passengers:")
print(females.head())


# Passengers older than 30
age_filter = df[df["Age"] > 30]

print("\nAge > 30:")
print(age_filter.head())


# Multiple conditions
rich_female = df[
    (df["Sex"] == "female") &
    (df["Fare"] > 100)
]

print("\nRich Female Passengers:")
print(rich_female)


# ==========================================
# 8. SORTING
# ==========================================

print("\nSort by Age:")
print(df.sort_values(by="Age").head())

print("\nHighest Fare:")
print(df.sort_values(by="Fare", ascending=False).head())


# ==========================================
# 9. HANDLE MISSING VALUES
# ==========================================

print("\nMissing Values:")
print(df.isnull().sum())


# Fill missing age with mean age
df["Age"] = df["Age"].fillna(df["Age"].mean())

print("\nMissing values after filling:")
print(df["Age"].isnull().sum())


# ==========================================
# 10. GROUP BY
# ==========================================

# Average age by gender
grouped = df.groupby("Sex")["Age"].mean()

print("\nAverage Age by Gender:")
print(grouped)


# Survival count by gender
survival = df.groupby("Sex")["Survived"].sum()

print("\nSurvival by Gender:")
print(survival)


# ==========================================
# 11. VALUE COUNTS
# ==========================================

print("\nPassenger Class Counts:")
print(df["Pclass"].value_counts())


# ==========================================
# 12. UNIQUE VALUES
# ==========================================

print("\nUnique Embarked Values:")
print(df["Embarked"].unique())


# ==========================================
# 13. STRING OPERATIONS
# ==========================================

print("\nUppercase Names:")
print(df["Name"].str.upper().head())


# ==========================================
# 14. CREATE NEW COLUMN
# ==========================================

# Age category
df["AgeCategory"] = df["Age"].apply(
    lambda x: "Child" if x < 18 else "Adult"
)

print("\nNew Column:")
print(df[["Name", "Age", "AgeCategory"]].head())


# ==========================================
# 15. CONDITIONAL FILTER
# ==========================================

survived = df[df["Survived"] == 1]

print("\nSurvived Passengers:")
print(survived.head())


# ==========================================
# 16. ITERATE ROWS
# ==========================================

print("\nIterating Rows:")

for index, row in df.head().iterrows():

    print(row["Name"], row["Fare"])


# ==========================================
# 17. SAVE CLEANED DATA
# ==========================================

df.to_csv("cleaned_titanic.csv", index=False)

print("\nCleaned CSV saved")


# ==========================================
# IMPORTANT PANDAS CONCEPTS
# ==========================================

"""
DataFrame:
2D table structure

Series:
Single column

loc:
Label-based indexing

iloc:
Index-based indexing

groupby:
Aggregation like SQL GROUP BY

apply():
Apply custom function

fillna():
Handle missing values
"""
```
